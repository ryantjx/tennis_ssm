```python
# Create a concise matches DataFrame with all relevant information
# Extract player details and match results clearly

matches_clean = matches_full.with_columns([
    # Extract player 1 (first player in the list)
    pl.col("players").list.get(0).struct.field("id").alias("player1_id"),
    pl.col("players").list.get(0).struct.field("name").alias("player1_name"),
    pl.col("players").list.get(0).struct.field("country").alias("player1_country"),
    
    # Extract player 2 (second player in the list)
    pl.col("players").list.get(1).struct.field("id").alias("player2_id"),
    pl.col("players").list.get(1).struct.field("name").alias("player2_name"),
    pl.col("players").list.get(1).struct.field("country").alias("player2_country"),
    
    # Convert start_date to timestamp
    pl.col("start_date").str.to_date().alias("date"),
    
    # Determine winner (1 if player1 won, 2 if player2 won)
    pl.when(pl.col("winner_id") == pl.col("players").list.get(0).struct.field("id"))
      .then(1)
      .when(pl.col("winner_id") == pl.col("players").list.get(1).struct.field("id"))
      .then(2)
      .otherwise(None)
      .alias("winner"),
    
    # Extract source information
    pl.col("sources").list.get(0).struct.field("key").alias("source_provider"),
    pl.col("sources").list.get(0).struct.field("source_match_id").alias("source_match_ref"),
]).select([
    # Tournament info
    pl.col("tour"),
    pl.col("name").alias("tournament"),
    pl.col("level").alias("tier"),
    pl.col("surface"),
    pl.col("year"),
    pl.col("date"),
    
    # Match info
    pl.col("round"),
    pl.col("discipline").alias("event_type"),
    pl.col("draw"),
    
    # Player 1
    pl.col("player1_id"),
    pl.col("player1_name"),
    pl.col("player1_country"),
    
    # Player 2
    pl.col("player2_id"),
    pl.col("player2_name"),
    pl.col("player2_country"),
    
    # Result
    pl.col("winner"),
    pl.col("score"),
    pl.col("status").alias("match_state"),
    
    # Source info
    pl.col("match_id").alias("source_match_id"),
    pl.col("id").alias("source_tournament_id"),
    pl.col("source_provider"),
    pl.col("source_match_ref"),
    pl.col("source_file"),
])

print(f"Clean matches DataFrame shape: {matches_clean.shape}")
print(f"\nColumns: {matches_clean.columns}")
matches_clean.head()
```
